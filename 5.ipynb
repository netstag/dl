```python
# If required, install the libraries in your Jupyter environment:
# %pip install -q gensim nltk

import re
import nltk
import gensim

from nltk.corpus import stopwords
from nltk.tokenize import sent_tokenize, word_tokenize
from gensim.models import Word2Vec

nltk.download("punkt", quiet=True)
nltk.download("punkt_tab", quiet=True)
nltk.download("stopwords", quiet=True)

print("Gensim version:", gensim.__version__)
print("Libraries imported successfully.")
```

```python
raw_text = '''
I am Aarav Mehta.
I am a Machine Learning student.
I am also an Artificial Intelligence student.
I study Machine Learning and Artificial Intelligence.
I am interested in learning how computers identify patterns from data.
Machine Learning uses data and algorithms to train predictive models.
Artificial Intelligence enables machines to perform tasks that usually require human intelligence.
I use Python programming to practice Machine Learning concepts.
Python helps me work with data, algorithms, and machine learning models.
I am learning about neural networks and deep learning.
I explore how models learn from examples and make predictions.
My practical work includes data preprocessing, model training, and evaluation.
I enjoy solving programming problems and improving my coding skills.
I want to build intelligent applications using Machine Learning and Artificial Intelligence.
My goal is to become skilled in data science, machine learning, and AI.
'''

print("Raw corpus:")
print(raw_text)
```

```python
# Step 1: Convert the raw text to lowercase and split it into sentences.
sentences = sent_tokenize(raw_text.lower())

# Step 2: Load English stop words.
stop_words = set(stopwords.words("english"))
```

```python
# Step 3: Clean and tokenize each sentence.
data = []

for sentence in sentences:
    # Remove punctuation, digits, and non-letter characters.
    cleaned_sentence = re.sub(r"[^a-z\s]", " ", sentence)

    # Remove extra spaces.
    cleaned_sentence = re.sub(r"\s+", " ", cleaned_sentence).strip()

    # Split the sentence into individual words.
    words = word_tokenize(cleaned_sentence)

    # Remove stop words and single-character tokens.
    words = [
        word for word in words
        if word.isalpha() and len(word) > 1 and word not in stop_words
    ]

    # Keep sentences that have at least two words after preprocessing.
    if len(words) >= 2:
        data.append(words)

print("Number of raw sentences:", len(sentences))
print("Number of prepared sentences:", len(data))

print("\nPrepared data (tokenized sentences):")
for i, words in enumerate(data, start=1):
    print(f"{i}. {words}")
```

```python
vocabulary = sorted({word for sentence in data for word in sentence})

print("Vocabulary size:", len(vocabulary))
print("Vocabulary:")
print(vocabulary)
```

```python
window_size = 2
training_examples = []

for sentence in data:
    for target_index in range(window_size, len(sentence) - window_size):
        context = (
            sentence[target_index - window_size:target_index]
            + sentence[target_index + 1:target_index + window_size + 1]
        )
        target = sentence[target_index]
        training_examples.append((context, target))

print("Number of context-target examples:", len(training_examples))
print("\nFirst 10 training examples:")
for context, target in training_examples[:10]:
    print("Context:", context, "=> Target:", target)
```

```python
window_size = 2
training_examples = []

for sentence in data:
    for i in range(window_size, len(sentence) - window_size):
        context = sentence[i-window_size:i] + sentence[i+1:i+3]
        target = sentence[i]

        training_examples.append((context, target))

print("Training examples:")

for example in training_examples[:10]:
    print(example)
```

```python
vector_size = 50

model = Word2Vec(
    sentences=data,
    vector_size=vector_size,
    window=window_size,
    min_count=1,
    workers=1,
    sg=0,          # 0 = CBOW, 1 = Skip-gram
    negative=5,
    epochs=300,
    seed=42
)

print("CBOW model trained successfully.")
print("Vocabulary size learned by model:", len(model.wv))
print("Vector size:", model.vector_size)
```

```python
print("Words learned by the model:")
print(list(model.wv.key_to_index.keys()))

example_word = "machine"
if example_word in model.wv:
    print("\nExample word:", example_word)
    print("Embedding vector (first 10 values):")
    print(model.wv[example_word][:10])
```

```python
example_word = "machine"

if example_word in model.wv:
    print("Input word:", example_word)
    print("\nTop similar words:")
    for similar_word, score in model.wv.most_similar(
        example_word, topn=min(10, len(model.wv) - 1)
    ):
        print(f"{similar_word:15s} similarity score = {score:.4f}")
else:
    print("The example word is not in the vocabulary.")
```

```python
word = input("Enter a word: ").lower().strip()

if word in model.wv:
    similar_words = model.wv.most_similar(word, topn=10)

    print("\nSimilar words for:", word)

    for w, score in similar_words:
        print(w, ":", round(score, 4))

else:
    print("Word not found in corpus.")
    print("Available words:")
    print(list(model.wv.key_to_index.keys()))
```
